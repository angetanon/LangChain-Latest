# Chat Model

```python
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(model="gpt-3.5-turbo-0125")
```

```python
messages = [
    ("system", "You are an historian expert"),
    ("human", "Tell me one curious fact about the Kennedy family"),
]

response = chatModel.invoke(messages)  
```

```python
print(response)
```
