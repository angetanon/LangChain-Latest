# Data Loader

```python
# Txt data loading
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data\\be-good.txt")
documents = loader.load()
documents
 
```

```python
# CSV loader

from langchain_community.document_loaders import CSVLoader

loader = CSVLoader("data\\Street_Tree_List.csv")
documents = loader.load()
#documents
```

```python
# HTML loader
from langchain_community.document_loaders import BSHTMLLoader

loader = BSHTMLLoader("data/100-startups.html", open_encoding="utf-8")
documents = loader.load()
documents
```

```python
# pdf loader

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("data\\5pages.pdf")
documents = loader.load_and_split()
documents
```

```python
documents[0].page_content
```

```python
# Wikipedia loader
import wikipedia
from langchain_community.document_loaders import WikipediaLoader

wikipedia.set_user_agent("LangChainLearningDemo/1.0")
loader = WikipediaLoader(query="Tesla", load_max_docs=1)
documents = loader.load()
document = documents[0].page_content
```

```python
document
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages([
    ("human", "Answer this {question}, here is some extract {context}"),
])

messages = chat_template.format_messages(
    name="Tesla", 
    question="Tell me about Tesla", 
    context=document
)
```

```python
response = chatModel.invoke(messages)
```
