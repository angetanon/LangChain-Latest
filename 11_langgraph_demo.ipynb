```python
import os
from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv())
openai_api_key = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-3.5-turbo-0125" )
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults

search_results = TavilySearchResults(max_results=2)
search_results.invoke("Who are the top stars of the 2026 world cup ?")
```

```python
tools = [search_results]
```

```python
tools
```

```python
from langchain.agents import create_agent

agent_executor = create_agent(llm, tools)
```

```python
from langchain_core.messages import HumanMessage

response = agent_executor.invoke({"messages": [HumanMessage(content="Who are the top stars of the 2026 world cup ?")]})

response["messages"]
```

```python
from langchain_core.messages import HumanMessage

response = agent_executor.invoke({"messages": [HumanMessage(content="What is python ?")]})

response["messages"]
```

```python
for chunk in agent_executor.stream({"messages": [HumanMessage(content="What is python ?")]}):
    print(chunk)
```

### Adding memory
* Adding memory in LangGraph is very similar to what we did with LangChain.

```python
from langgraph.checkpoint.memory import MemorySaver
memory = MemorySaver()
```

```python
agent_executor.memory = create_agent(llm, tools, checkpointer=memory)

config = {"configurable": {"thread_id": "001"}} # Configuration for the agent's memory thread: user 001
```

```python
for chunk in agent_executor.stream({"messages": [HumanMessage(content="Who won the 2026 soccer world cup ?")]}):
    print(chunk)
    print("------")
```

```python
for chunk in agent_executor.stream({"messages": [HumanMessage(content="Who were the top stars of that winner team ?")]}):
    print(chunk)
    print("------")
```

```python
config = {"configurable": {"thread_id": "002"}} # Configuration for the agent's memory thread: user 002
for chunk in agent_executor.stream({"messages": [HumanMessage(content="About what soccer team we were talking about ?")]}):
    print(chunk)
    print("------")
```
